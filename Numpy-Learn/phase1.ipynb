## Numpy array and Basics

```python
import numpy as np
```

## Creating array with list

```python
arr_1d = np.array([1,2,3,4,5])
print("1d array:", arr_1d)

arr_2d = np.array([[2,3,4,5,6],[7,8,9,0,1]])

print("2d array:",arr_2d)
```

### List vs Numpy Array

```python
py_list = [1,2,3,4]
print("Python list multiplication:", py_list * 2)
```

```python
np_array = np.array([1,2,3,4]) #Element wise multiplication 
print("Python list multiplication:", np_array * 2)
```

```python
import time
start = time.time()
py_list = [i*2 for i in range(100000)]
print("\n List operation time", time.time() - start)

start = time.time()
np_array = np.arange(100000) * 2
print("\n Numpy operation time", time.time() - start)
```

### Creating array from Scratch

```python
zeros = np.zeros((3,4))
print("Zeros array =\n ",zeros)

ones = np.ones((3,4))
print("Ones array =\n ",ones)

full = np.full((2,2), 7)
print("Full array =\n ",full)

random = np.random.random((2,3))
print("Random array =\n ",random)

sequence = np.arange(0,10,2) # Ranges from 0 to 9 with step =2.
print("Sequence array =\n ",sequence)
```

### Vector, Matrices and Tensor

```python
vector = np.array([1,2,3])
print("Vector:",vector)

matrices = np.array([[1,2,3],
                    [4,5,6]])
print("\nMatrices:\n", matrices)

tensor = np.array([[[1,2], [3,4]],
                  [[5,6],[7,8]]])

print("\nTensor:",tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],
               [4,5,6]]) #Numpy array's optimised as they store similar data of types as we can do arr = np.array([[1,2,3],
               # [4,5,True]]) as well.
print("Shape of array :", arr.shape) # Gives rows and column
print("Dimension:",arr.ndim) # Gives Dimension
print("Size:",arr.size) # Gives Size(Total number of elements)
print("Dtype:",arr.dtype) # Gives Datatype
```

### Array Reshaping

```python
arr = np.arange(12)
print("Original Array :", arr)
```

```python
reshaped = arr.reshape((3,4)) # Convert into 3*4 matrix
print("\n Reshaped array:", reshaped)

flattened= reshaped.flatten()
print("\n flattened array:",flattened) #Convert into flat array in one row.

raveled = reshaped.ravel() # It returns the view not copy.
print("\n Raveled array :", raveled) 

transpose = reshaped.T
print("\n Transpose:",transpose) # It returns the transpose of the matrix.
```
