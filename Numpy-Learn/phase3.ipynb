## Advance operation with Business example

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Data Structure : [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 120000, 140000, 160000, 190000], # Beijing Bites
    [3, 200000, 230000, 250000, 300000], # Pizza Hub
    [4, 180000, 210000, 240000, 270000], # Burger Point
    [5, 160000, 185000, 205000, 230000]  # Chai Point
])

print("=== Zomato Sales Analysis ===")
print("\nSales data type:",sales_data.shape)
print("\n Sample data for 1st 3 restaurant:\n", sales_data[:3])
```

```python
# Total sales per year
print(np.sum(sales_data, axis = 0))
yearly_total = np.sum(sales_data[:, 1:], axis =0)
print(yearly_total)

# Minimum sales per restaurant

min_sales = np.min(sales_data[:, 1:], axis=1 )
print(min_sales)

# Maximum sales per year

max_sales = np.max(sales_data[:,1:], axis=0)
print(max_sales)


# Average sales per restaurant

avg_sales = np.mean(sales_data[:, 1:], axis= 1)
print(avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:,1:], axis =1)
print(cumsum)
plt.Figure(figsize=(6,6))
plt.plot(np.mean(cumsum,axis = 0))
plt.title("Average Cumulative Sales across all Restaurant ")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("Vector Addition:", vector1+vector2)

print("\n Multiplication of vector:", vector1 * vector2 )

print("Dot Product:", np.dot(vector1, vector2))
angle = np.arccos(np.dot(vector1, vector2)/ (np.linalg.norm(vector1) * np.linalg.norm(vector2))) # Finding out the angle between two vectors
print("Angle is :", angle)



#Vectorised Operation

restaurant_types = np.array(['biryani', 'chinese', 'pizza', 'burger', 'cafe'])

vectorised_upper = np.vectorize(str.upper)
print("Vectorized Upper: ", vectorised_upper(restaurant_types))

```
